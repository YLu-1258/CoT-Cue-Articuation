```python
import pandas as pd
```

```python
df = pd.read_json("data/stanford_gsm8k_responses/biased/stanford_professor_responses.jsonl", lines=True)
```

```python
df
```

```python
# Load and sort the unbiased responses
unbiased_df = pd.read_json("data/gsm8k/unbiased_responses/gsm8k_test_responses.jsonl", lines=True)
unbiased_df_sorted = unbiased_df.sort_values('question_id')
unbiased_df_sorted.to_json("data/gsm8k/unbiased_responses/gsm8k_test_responses.jsonl", orient='records', lines=True)
```

```python
# Check that every question id exists in both datasets
print("Checking question IDs in both datasets...")

# Get unique question IDs from both datasets
biased_ids = set(df['question_id'].unique())
unbiased_ids = set(unbiased_df_sorted['question_id'].unique())

print(f"Biased dataset has {len(biased_ids)} unique question IDs")
print(f"Unbiased dataset has {len(unbiased_ids)} unique question IDs")

# Check for missing IDs in biased dataset
missing_in_biased = unbiased_ids - biased_ids
if missing_in_biased:
    print(f"❌ Missing in biased dataset: {sorted(missing_in_biased)}")
else:
    print("✅ All unbiased question IDs present in biased dataset")

# Check for missing IDs in unbiased dataset  
missing_in_unbiased = biased_ids - unbiased_ids
if missing_in_unbiased:
    print(f"❌ Missing in unbiased dataset: {sorted(missing_in_unbiased)}")
else:
    print("✅ All biased question IDs present in unbiased dataset")

# Check for complete overlap
if biased_ids == unbiased_ids:
    print("✅ Perfect match: Both datasets have identical question IDs")
    print(f"Range: {min(biased_ids)} to {max(biased_ids)}")
else:
    print("❌ Question ID sets do not match perfectly")
    
# Check for expected range (0 to 999 for GSM8K test set)
expected_ids = set(range(1000))
if biased_ids == expected_ids:
    print("✅ Biased dataset has complete 0-999 range")
else:
    print(f"❌ Biased dataset missing: {sorted(expected_ids - biased_ids)}")
    
if unbiased_ids == expected_ids:
    print("✅ Unbiased dataset has complete 0-999 range") 
else:
    print(f"❌ Unbiased dataset missing: {sorted(expected_ids - unbiased_ids)}")
```
